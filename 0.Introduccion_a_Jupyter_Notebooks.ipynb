## Ejecución de tutoriales

Los cuadernos de Jupyter son documentos que combinan código Python, texto enriquecido y código adicional (como Bash), organizados en bloques ejecutables. Estos cuadernos permiten ejecutar los bloques en desorden y varias veces, lo que ofrece un mayor control sobre lo que se está ejecutando en comparación con un script de Python. Para ejecutarlos, está instalada la aplicación Jupyter Notebook, que también incluye funciones de explorador de archivos y editor de texto.

A continuación, se muestra un ejemplo de un bloque de Python. Primero, selecciona el bloque haciendo clic sobre él. Debería aparecer un borde verde o azul alrededor del bloque, indicando que está seleccionado. A continuación, pulsa el botón `>| Run` en la barra de herramientas de la parte superior de la pantalla. El siguiente bloque debería quedar ahora resaltado.

```python
i = 0
```

El estado se conserva entre los distintos bloques de Python. Prueba a ejecutar el siguiente bloque varias veces para ver cómo cambia el número impreso (con `Control+Enter` se ejecuta el bloque sin pasar al siguiente):

```python
i += 1
print(i)
```

Los cuadernos también permiten ejecutar bloques especiales. Dos de los bloques especiales utilizados a lo largo de los tutoriales de ChipWhisperer son los bloques de `Bash`, que permiten ejecutar comandos de Bash (nota: si no estás utilizando una terminal tipo Bash para ejecutar Jupyter, estos bloques no te funcionarán):

```python
%%bash
echo "Test bash"
i=0
for j in *; do let "i=i+1"; done
echo "there are" $i "files/folders in this directory"
```

Y los bloques `run`, que ejecutan otros cuadernos:

```python
%run "Helper_Scripts/Test_Run.ipynb"
```

Los bloques de código también se pueden editar haciendo clic dentro de la parte de código del bloque (el área gris). Sabrás si estás en el modo de edición por la presencia de un borde verde alrededor del bloque. Si no estás en el modo de edición, el borde será azul.

## Creación de gráficos con matplotlib

En primer lugar: crear gráficos con matplotlib. Es la opción más fácil de usar, ya que puedes representar datos con `plt.plot()` y también definir un color, como se muestra a continuación:

    import matplotlib.pylab as plt
    plt.plot([1,2,2,4,5], 'r')
    
Si no ves el resultado, añade la siguiente sentencia al final:

    plt.show()
    

```python
import matplotlib.pylab as plt
plt.plot([1,2,2,4,5], 'r')
plt.show()
```

A menudo resulta útil tener un gráfico interactivo. Podemos lograrlo con un comando especial o *magic* al inicio de nuestro cuaderno, que se ve así:

    %matplotlib notebook
    import matplotlib.pylab as plt
    
    plt.plot([1,2,2,4,5], 'r')
    plt.plot([3,1,5,5,7], 'g')
    
Pruébalo justo aquí:

```python
%matplotlib widget
import matplotlib.pylab as plt
    
plt.plot([1,2,2,4,5], 'r')
plt.plot([3,1,5,5,7], 'g')
```

Otra función que a menudo necesitarás es la capacidad de ejecutar comandos de graficado desde dentro de un bucle; por ejemplo, para representar datos en tiempo real.

Esto puede ser un poco lento, pero es una herramienta bastante potente. Se logra fácilmente usando el gráfico interactivo (visto antes) en una celda y luego realizando la actualización del gráfico en otra celda.

Es probable que necesites añadir un `fig.canvas.draw()` dentro del bucle para forzar la actualización. A continuación se muestra un ejemplo de cómo trazar una serie de puntos (¡qué emocionante!) con un retraso entre cada trazado. Ten en cuenta que la figura se genera en el primer bloque y la actualización real ocurre en el segundo bloque.

```python
%matplotlib widget
import matplotlib.pylab as plt
fig = plt.figure()
```

```python
import time

for i in range(0, 10):
    time.sleep(1)
    
    ## to update the plot:
    plt.plot(-5, i, '.')
    fig.canvas.draw()
```

## Widgets

Otra característica superútil son los llamados «widgets», que permiten añadir elementos interactivos al cuaderno. Resultan excelentes para proporcionar información de estado (como el número de trazas capturadas) dentro de un bucle.

Consulta la [documentación de los widgets](https://ipywidgets.readthedocs.io/en/latest/examples/Widget%20Basics.html) para conocer todos los detalles sobre cualquier widget que puedas necesitar. Puedes mostrarlos de forma sencilla como se indica a continuación:

```python
import ipywidgets as widgets

slider_widget = widgets.FloatSlider( value=0.0,
                                     min=0,
                                     max=1.0,
                                     step=0.01,
                                     description="Doneness:",
                                     disabled=True,
                                     continuous_update=False,
                                     orientation='horizontal',
                                     readout=True,
                                     readout_format='.01f')
                                            
status_widget = widgets.IntText(     value=0,
                                     description="Iteration No",
                                     disabled=True)

display(slider_widget)
display(status_widget)
```

```python
import time
for i in range(0, 12):
    slider_widget.value = i / 12.0
    status_widget.value = i
    time.sleep(0.5)
```
